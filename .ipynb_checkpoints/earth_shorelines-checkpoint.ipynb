```python
lake = gdf[gdf["Lake_name"] == "Ontario"]
```

```python
lake = gdf[gdf["Hylak_id"] == 184967]
```

```python
geom = lake.geometry.iloc[0]

coords = geom.exterior.coords  # polygon boundary
coords = list(coords)

x = [c[0] for c in coords]
y = [c[1] for c in coords]
```

```python
import matplotlib.pyplot as plt

plt.plot(x, y)
plt.axis("equal")
plt.title("Lake Ontario coastline")
plt.show()
```
