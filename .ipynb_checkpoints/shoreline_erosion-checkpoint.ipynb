```python
import numpy as np
import matplotlib.pyplot as plt
import geopandas as gpd
from shapely.ops import unary_union, linemerge

# coastline: x-position as a function of y
n = 100
coast = np.random.rand(n) * 10  # rough "young" coastline

k = 0.1       # erosion coefficient
dt = 0.1      # time step
steps = 300   # number of iterations

gdf = gpd.read_file("C:/Users/isabe/Downloads/HydroLAKES_polys_v10_shp/HydroLAKES_polys_v10_shp/HydroLAKES_polys_v10.shp")
```

```python
lake = gdf[gdf["Lake_name"] == "Lake Powell"]
geom = lake.geometry.iloc[0]

coords = geom.exterior.coords  # polygon boundary
coords = list(coords)

x = [c[0] for c in coords]
y = [c[1] for c in coords]

k=0.1
dt=0.1
steps=300
n=len(x)

print(len(x), len(y))
```

```python
def erode_closed_shoreline(x, y, alpha=0.05, steps=500):
    """
    Simple curvature-smoothing erosion model for a closed shoreline.
    alpha controls erosion strength per step.
    """
    x = np.array(x, dtype=float)
    y = np.array(y, dtype=float)
    history = []

    for t in range(steps):
        if t % 10 == 0:
            history.append((x.copy(), y.copy()))

        # neighbors, with wraparound because lake is closed
        x_prev = np.roll(x, 1)
        x_next = np.roll(x, -1)
        y_prev = np.roll(y, 1)
        y_next = np.roll(y, -1)

        # curvature-like smoothing
        x = x + alpha * (x_prev + x_next - 2*x)
        y = y + alpha * (y_prev + y_next - 2*y)

    history.append((x.copy(), y.copy()))
    print(len(x), len(y))
    return x, y, history


```

```python
x_eroded, y_eroded, history = erode_closed_shoreline(x, y, alpha=0.03, steps=500)
```

```python
plt.figure(figsize=(20,20))

for i, (xs, ys) in enumerate(history):
    plt.plot(xs, ys, label=f"step {i*10}")

plt.axis("equal")
plt.title("Full lake shoreline erosion model")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# x, y should already be your lake shoreline arrays
x = np.array(x, dtype=float)
y = np.array(y, dtype=float)

k = 0.1
dt = 0.5
steps = 300

n = len(x)

# wave energy around the shoreline
wave_energy = 1 + 0.5 * np.sin(np.linspace(0, 4*np.pi, n))

fig, ax = plt.subplots(figsize=(6,6))
line, = ax.plot(x, y)

ax.set_aspect("equal")
ax.set_xlim(x.min(), x.max())
ax.set_ylim(y.min(), y.max())

def update(frame):
    global x, y

    lap_x = np.roll(x, -1) - 2*x + np.roll(x, 1)
    lap_y = np.roll(y, -1) - 2*y + np.roll(y, 1)

    x = x + k * wave_energy * lap_x * dt
    y = y + k * wave_energy * lap_y * dt

    line.set_data(x, y)
    ax.set_title(f"Step {frame}")

    return line,

ani = FuncAnimation(fig, update, frames=steps, interval=50)

ani.save("_earth_lake_shoreline_erosion.gif", writer="pillow", fps=20)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

n = 300
coast = np.random.rand(n) * 10

k = 0.1
dt = 0.1
steps = 300

wave_energy = 1 + 0.5*np.sin(np.linspace(0, 4*np.pi, n))

fig, ax = plt.subplots()
line, = ax.plot(coast)
ax.set_ylim(0, 10)

def update(frame):
    global coast
    laplacian = np.roll(coast, -1) - 2*coast + np.roll(coast, 1)
    coast = coast + k * wave_energy * laplacian * dt
    
    line.set_ydata(coast)
    ax.set_title(f"Step {frame}")
    return line,

ani = FuncAnimation(fig, update, frames=steps, interval=50)
plt.show()

ani.save("coastline_erosion.gif", writer="pillow", fps=20)
```
